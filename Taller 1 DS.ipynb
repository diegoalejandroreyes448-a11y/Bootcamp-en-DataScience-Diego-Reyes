## Variables: Crea un programa que permita interactuar con el usuario para recolectar datos. Asegurate que los datos recopilados tengan el formato correcto para poder hacer cálculos.

1. Solicita al usuario ingresar el nombre de un producto.
2. Pídele que ingrese el precio del producto y la cantidad vendida.
3. Convierte el precio a un número decimal (float) y la cantidad a un número entero (int).
4. Calcula el total de la venta y muestra un mensaje en pantalla con el resumen.

```python
# Primer paso, solicitamos los datos del producto al usuario
nombre_producto = input("Ingresa el nombre del producto: ")
precio_input = input("Ingresa el precio del producto: ")
cantidad_input = input("Ingresa la cantidad vendida: ")

# Segundo paso, convertimos los datos al tipo numérico correcto
precio = float(precio_input)
cantidad = int(cantidad_input)

# Tercer paso, calculamos el total de la venta
total_venta = precio * cantidad

# Cuarto paso, mostramos el resultado al usuario
print("Total de la venta:")
print(f"Producto: {nombre_producto}")
print(f"Precio unitario: ${precio}")
print(f"Cantidad vendida: {cantidad}")
print(f"Total de la venta: ${total_venta}")
```

## Estructuras de Datos (Listas, Tuplas, Diccionarios)

Crea un programa para organizar información de múltiples ventas utilizando las estructuras adecuadas según la naturaleza de los datos.

Tupla: Crea una tupla con las categorías de productos permitidas en la tienda (ej. "Electrónica", "Ropa", "Alimentos"). Se usa una tupla porque las categorías de la tienda no cambian todos los días.
Diccionario: Crea un diccionario que represente una venta específica. Debe tener las llaves: producto, categoria, precio, cantidad y total.
Lista: Crea una lista vacía llamada registro_ventas y añade el diccionario de la venta que acabas de crear.

```python
# Tupla con las categorías permitidas en la tienda
categorias_permitidas = ("Electrónica", "Ropa", "Alimentos")
print("Categorías disponibles:", categorias_permitidas)

# Diccionario que representa una venta específica
venta_especifica = {
    "producto": nombre_producto,
    "categoria": categorias_permitidas[0], 
    "precio": precio,
    "cantidad": cantidad,
    "total": total_venta
}

# Lista que almacena el historial de ventas
registro_ventas = []
registro_ventas.append(venta_especifica)

print("Registro de ventas:", registro_ventas)
```

## Estructuras de Control (Condicionales y Ciclos)

Haz un programa que ayude a automatizar procesos repetitivos y aplicar reglas de negocio (lógica condicional).

Crea un ciclo for que recorra tu lista de registro_ventas (puedes agregar un par de ventas manuales extra al código para hacer la prueba).
Dentro del ciclo, aplica un condicional if: Si el total de la venta es mayor a $100000, aplica un descuento del 10%.
Imprime el reporte final mostrando el producto, el total original y el total con descuento (si aplica).

```python
# Se agregan dos ventas adicionales al registro para probar la regla de descuento
venta_2 = {
    "producto": "Buzo",
    "categoria": "Ropa",
    "precio": 70000.0,
    "cantidad": 3,
    "total": 210000.0
}

venta_3 = {
    "producto": "Hamburguesa",
    "categoria": "Alimentos",
    "precio": 15000.0,
    "cantidad": 5,
    "total": 75000.0
}

registro_ventas.append(venta_2)
registro_ventas.append(venta_3)

# ciclo for para recorrer el registro de ventas y aplicar la regla de descuento
print("Reporte final de ventas")
for venta in registro_ventas:
    total_original = venta["total"]

    # Si el total supera $100.000, aplicamos un descuento del 10%
    if total_original > 100000:
        descuento = total_original * 0.10
        total_con_descuento = total_original - descuento
        print(f"Producto: {venta['producto']} | Total original: ${total_original} | "
              f"Descuento aplicado (10%): ${descuento} | Total con descuento: ${total_con_descuento}")
    else:
        print(f"Producto: {venta['producto']} | Total original: ${total_original} | Sin descuento")
```

## Funciones y Manejo de Errores

Crea código reutilizable y a prueba de errores de usuario siguiendo estos parámetros.

Crea una función llamada calcular_ingreso que reciba dos parámetros: precio y cantidad.
Dentro de la función, utiliza un bloque try-except. Intenta multiplicar el precio por la cantidad.
Si ocurre un error (por ejemplo, si en lugar de un número llega un texto), el except debe atraparlo, imprimir "Error: Los valores deben ser numéricos" y retornar 0.
Prueba tu función enviando valores correctos y luego enviando un texto para ver cómo se maneja el error.

```python
# Se crea la función para calcular el ingreso total de una venta, con manejo de errores
def calcular_ingreso(precio, cantidad):
    try:
        ingreso = precio * cantidad
        return ingreso
    except TypeError:
        print("Error: Los valores deben ser numéricos")
        return 0

# Prueba 1: valores numéricos correctos
resultado_1 = calcular_ingreso(1000, 3.5)
print("Resultado 1: ", resultado_1)

# Prueba 2: enviamos un texto en lugar de un número para activar el except, se coloca 3.5 para forzar el error
resultado_2 = calcular_ingreso("mil ", 3.5)
print("Resultado 2: ", resultado_2)
```
